```python

# Overall Business Performance

total_sales = df['Sales'].sum()
total_profit = df['Profit'].sum()
total_quantity = df['Quantity'].sum()
average_discount = df['Discount'].mean()

print("===== BUSINESS PERFORMANCE =====")

print("Total Sales: $", round(total_sales, 2))
print("Total Profit: $", round(total_profit, 2))
print("Total Quantity Sold:", total_quantity)
print("Average Discount:", round(average_discount * 100, 2), "%")
```

```python

# Sales by Product Category

category_sales = df.groupby('Category')['Sales'].sum()

print(category_sales.sort_values(ascending=False))
```

```python

# Extract year and month from Order Date

df['Year'] = df['Order Date'].dt.year
df['Month'] = df['Order Date'].dt.month

print(df[['Order Date', 'Year', 'Month']].head())
```

```python

# Calculate total sales for each month

monthly_sales = df.groupby('Month')['Sales'].sum()

print(monthly_sales)
```

```python

plt.figure(figsize=(10, 5))

monthly_sales.plot(
    kind='line',
    marker='o',
    title='Monthly Sales Analysis'
)

plt.xlabel("Month")
plt.ylabel("Total Sales ($)")
plt.xticks(range(1, 13))
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python

# Group data by discount level

discount_analysis = df.groupby('Discount').agg(
    Total_Sales=('Sales', 'sum'),
    Average_Profit=('Profit', 'mean'),
    Total_Profit=('Profit', 'sum')
)

print(discount_analysis)
```

```python

plt.figure(figsize=(9, 5))

plt.scatter(
    df['Discount'],
    df['Profit'],
    alpha=0.4
)

plt.title('Relationship Between Discount and Profit')
plt.xlabel('Discount')
plt.ylabel('Profit ($)')
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python

# Outlier detection using IQR

Q1 = df['Sales'].quantile(0.25)
Q3 = df['Sales'].quantile(0.75)

IQR = Q3 - Q1

lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

outliers = df[
    (df['Sales'] < lower_limit) |
    (df['Sales'] > upper_limit)
]

print("Lower Limit:", lower_limit)
print("Upper Limit:", upper_limit)
print("Number of Sales Outliers:", len(outliers))
```

# Project Conclusion – E-Commerce Sales Analysis

## Project Overview

The objective of this project was to analyse an e-commerce sales dataset using Python and Exploratory Data Analysis (EDA). The analysis focused on identifying sales trends, profitable product categories, regional performance and data quality.

## Tools and Technologies

* Python
* Pandas
* NumPy
* Matplotlib
* Seaborn
* Google Colab
* Kaggle

## Key Findings

1. The dataset contains 9,994 sales records.
2. Total sales generated were $2,297,200.86.
3. Total profit was $286,397.02.
4. Technology was the highest-performing category in both sales and profit.
5. The West region generated the highest sales.
6. November recorded the highest aggregated monthly sales.
7. No missing values or duplicate records were found.
8. A total of 1,167 sales outliers were identified using the IQR method.

## Conclusion

The analysis shows that Technology is a major contributor to the store's sales and profitability, while the West region is its strongest sales region.

November recorded the highest aggregated sales across the dataset's years. The outlier analysis also identified unusual sales transactions that may require further investigation.

This project provided practical experience in data preprocessing, exploratory data analysis, statistical calculations and data visualization using Python.
